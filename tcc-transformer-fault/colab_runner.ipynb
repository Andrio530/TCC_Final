```python
import os
import zipfile
from google.colab import files

BASE = "/content/tcc-transformer-fault"

# Os 3 arquivos gerados agora pelo script de multi-horizonte
arquivos = [
    "reports/metrics/lstm_multi_horizon_results_norm.json",
    "reports/figures/lstm_multi_horizon_r2_norm.png",
    "reports/figures/lstm_multi_horizon_mae_pct_norm.png"
]

zip_path = "/content/resultados_horizontes_finais.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
    for rel_path in arquivos:
        full_path = os.path.join(BASE, rel_path)
        if os.path.exists(full_path):
            zf.write(full_path, arcname=rel_path)
            print(f"✅ Adicionado: {rel_path}")
        else:
            print(f"⚠️  NÃO ENCONTRADO: {rel_path}")

print(f"\n📦 ZIP criado com sucesso: {zip_path}")
files.download(zip_path)
```
